## Soluzione esercizio: il Sor Franco e la domenica dei turisti

#### Descrizione agente (PEAS)

**Performance Measure**: Velocità, completezza e pieno carburante: si verifica quanto l'agente sia veloce a trovare un percorso ottimale unito alla necessità di completare tutti gli obiettivi (le corse da fare) senza che rimanga a corto di benzina

**Enviorment**: Grliglia 10x10: ambiente completamente osservabile, statico e deterministico

**Actuators**: celle lungo la grliglia attraverso il quale l'agente può navigare sulla mappa

**Sensors**: celle adiacenti e celle indicanti gli isolati e le zone di arrivo (luoghi, benzinaio, ...). L'agente sa dove non può passare e dove deve andare nel caso dovesse prendere i turisti oppure fare rifornimento

<br>

**Agente**: l'agente è un agente con obiettivo in quanto sa esattamente come e strutturato tutto l'ambiente ed è pienamente al corrente di tutte le possibili azioni e presenta più obiettivi: portare a destinazione ogni turista

<br>

**Stati**
* Iniziale: cella in posizione $P_{x=6, y=4, v=0}$ (v=quale viaggio sta facendo, se il primo o altri) e carburante al livello massimo
* generico: Posizione casuale sulla griglia $P_{x,y,v}$ e livello carburante Carb > 0
* finale: cella in posizione $P_{x=1, y=2, v=3}$. Questo stato si verifica solamente quando si arriva in quella cella con l'ultimo viaggio.

<br>

**Percezioni**
L'agente muovendosi nell'ambiente ed essendo questo completamente osservabile ottiene come percezione quella intera dell'ambiente con tutti i luoghi dove può e non può andare e dove deve andare.
