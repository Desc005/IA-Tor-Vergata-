## Esercizio 1 · Modellazione di una argomentazione

> *Se Attila morì, egli morì quando era vivo o morì mentre era morto. Ma non morì mentre era vivo, poiché la stessa persona sarebbe sia viva che morta. E non morì neppure mentre era morto, altrimenti sarebbe morto due volte. Dunque, Attila non morì.*

**Domanda:**

- Definire nel linguaggio del Calcolo Proposizionale la argomentazione definita dal testo precedente.
- Definire dunque tutte le proposizioni necessarie, e le loro composizioni definite dalle frasi del testo.
- Utilizzare un <span style="color:#d4a017">***qualsiasi metodo di inferenza***</span> che determini se la affermazione finale è o no conseguenza logica delle assunzioni iniziali.

**Risposta**

**Proposizioni**

M = Attila è morto

V = Attila è vivo

MV = Attila è morto quando era vivo

MM = Attila è morto quando era morto

T = morto due volte

-------

**Formule**

$M \implies (MM \lor MV)$

$MV \implies (M \land V)$

$MM \implies T$

$\neg T$

$\neg(M \land V)$

$V = \neg M$

------

**KB** = {$M \implies (MM \lor MV),  MV \implies (M \land V),  MM \implies T,  \neg T,  \neg(M \land V),  V = \neg M$}

**KB** $\models \alpha$

$\alpha = V$ 

-------

**Dimostrazione con metodi di inferenza**

Sappiamo che $M \implies (MM \lor MV)$, ma:

* Essendo vera la formula $MV \implies (M \land V)$, per contrapposizone si ottiene che $\neg (M \land V) \implies \neg MV$ anche è vera, applicando il Modus Ponens:

$$\frac{\neg (M \land V) \implies \neg MV \qquad \neg (M \land V)}{\neg MV}$$

Da qui si ottiene quindi che $\neg MV$ è vera

* Essendo vera la formula $MM \implies T$, per contrapposizione si ottiene che $\neg T \implies \neg MM$ anche è vera, applicando il Modus Ponens:

$$\frac{\neg T \implies \neg MM \qquad \neg T}{\neg MM}$$

Si ottiene che $\neg MM$ è vera.

* Essendo vera la formula $M \implies (MM \lor MV)$, per contrapposizione si ottiene che $\neg (MM \lor MV) \implies \neg M$ anche è vera ed essendo sia $\neg MM$ e sia $\neg MV$ vere, anche $\neg MM \land \neg MV$ è vera. Poiché quella formula è vera, si può ottenere, applicando De Morgan, la formula $\neg (MM \lor MV)$. Applicando il Modus Ponens:

$$\frac{\neg (MM \lor MV) \implies \neg M \qquad \neg (MM \lor MV)}{(\neg M)}$$

Si ottiene che $\neg M$ è vera.

Di conseguenza, essendo $V = \neg M$, si ottiene che V è vera e quindi $\alpha = V$ è conseguenza logica di KB.

## Esercizio 2 · Ancora una modellazione in CProp

Definire in Logica Proposizionale la seguente argomentazione:

> a) *Gianni è scapolo.*
>
> b) *Dunque, Gianni non è sposato.*

- Data una traduzione in CProp di a) e b), dire se la argomentazione è valida, cioè se la tesi b) consegue logicamente dalla ipotesi o no.
- Discutere le ragioni della risposta.

**Risposta**

p1: Gianni è scapolo

p2: Gianni è sposato

$p1 \implies \neg p2$

L'argomentazione non è valida in logica proposizionale in quanto la tesi $\neg p2$ non è detto che consegua logicamente dall'ipotesi *p1*, questo perché in logica proposizionale si valuta il risultato di una formula solamente in base ai valori assegnati ad ogni elemento che la costituisce senza considerare il significato reale. In questo caso, senza la presenza di un'ulteriore ipotesi che stabilisce che se si è scapoli non si può essere sposati stabilendo quindi che le due proposizioni debbano avere valori diversi, nulla vieta di assegnare a p1 e a p2 entrambi Vero. Questo renderebbe la formula $p1 \implies \neg p2$ falsa e dimostrerebbe che la tesi non consegue dall'ipotesi.

Stato iniziale:

$Init^0 \equiv A^{0}_{5,sx} \land \neg B^{0}_{5,sx} \land B^{0}_{5,dx}$

<br>

Goal:

$Goal^t \equiv \neg B^{t}_{5,sx} \land \neg B^{t}_{5,dx}$

<br>

Transition 1:

l'unico movimento che si può fare per il raggiungimento del goal è muoversi verso destra, quindi si impone $Destra^0 = vero$

$(A^{t+1}_{5,dx} \Leftrightarrow ((A^{t}_{5,sx} \land Destra^t) \lor (A^{t}_{5, dx} \land \neg Sinistra^t))) \equiv (((A^{t}_{5,sx} \land Destra^t) \lor (A^{t}_{5, dx} \land \neg Sinistra^t)) \implies A^{t+1}_{5,dx}) \land (A^{t+1}_{5,dx} \implies ((A^{t}_{5,sx} \land Destra^t) \lor (A^{t}_{5, dx} \land \neg Sinistra^t)))$

essendo nella KB, l'assioma è vero, quindi si prende una delle due parti dell'AND per la dimostrazione (essendo enrambe le parti vere)

$((A^{0}_{5,sx} \land Destra^0) \lor (A^{0}_{5, dx} \land \neg Sinistra^0)) \implies A^{1}_{5,dx}$

Avendo:

* $A^{0}_{5,sx}$ = Vero da $Init^0$

* $Destra^0$ = Vero per supposizione

Allora:

$((A^{0}_{5,sx} \land Destra^0) \lor (A^{0}_{5,dx} \land Sinistra^0))$ = vero

con modus ponens si ottiene:

$\frac{((A^{0}_{5,sx} \land Destra^0) \lor (A^{0}_{5, dx} \land \neg Sinistra^0)) \implies A^{1}_{5,dx} \qquad ((A^{0}_{5,sx} \land Destra^0) \lor (A^{0}_{5,dx} \land Sinistra^0))}{A^{1}_{5,dx}}$

quindi *$A^{1}_{5,dx}$ è vero*

<br>

Come effettuato in precedenza, da:

$(B^{t+1}_{5,sx} \Leftrightarrow (B^{t}_{5,sx} \land \neg (A^{t}_{5, sx} \land Aspira^t))) \equiv (((B^{t}_{5,sx} \land \neg (A^{t}_{5, sx} \land Aspira^t)) \implies B^{t+1}_{5,sx}) \land (B^{t+1}_{5,sx} \implies (B^{t}_{5,sx} \land \neg (A^{t}_{5, sx} \land Aspira^t))))$

per contraddizione di:

$B^{t+1}_{5,sx} \implies (B^{t}_{5,sx} \land \neg (A^{t}_{5, sx} \land Aspira^t))$

si ottiene:

$\neg (B^{t}_{5,sx} \land \neg (A^{t}_{5, sx} \land Aspira^t)) \implies \neg B^{t+1}_{5,sx}$

Con $B^{0}_{5,sx}$ = falso da $Init^0$ si ottiene che:

$B^{0}_{5,sx} \land \neg (A^{0}_{5, sx} \land Aspira^0)$ = falso

essendo nella contraddizione precedente negata, si ottiene che la formula diventa *vera*

Essendo la formula vera, per modus ponens:

$\frac{\neg (B^{0}_{5,sx} \land \neg (A^{0}_{5, sx} \land Aspira^0)) \implies \neg B^{1}_{5,sx} \qquad \neg (B^{0}_{5,sx} \land \neg (A^{0}_{5, sx} \land Aspira^0))}{\neg B^{1}_{5,sx}}$

Si ottiene che *$\neg B^{1}_{5,sx}$ è vera*

<br>

Effettuando le trasformazioni delle equivalenze logiche fatte in precedenza da:

$B^{t+1}_{5,dx} \Leftrightarrow (B^{t}_{5, dx} \land \neg (A^{t}_{5, dx} \land Aspira^t))$

si ottiene:

$(B^{t}_{5, dx} \land \neg (A^{t}_{5, dx} \land Aspira^t)) \implies B^{t+1}_{5,dx}$

Avendo:

* $B^{0}_{5,dx}$ = Vero da $Init^0$

* $A^{0}_{5,dx}$ = Falso, in quanto l'aspirapolvere al tempo 0 si trova nella stanza di sinistra e dalle regole non si può trovare nell'altra stanza

si ottiene:

$(B^{0}_{5, dx} \land \neg (A^{0}_{5, dx} \land Aspira^0))$ = Vero

applicando il modus ponens:

$\frac{(B^{0}_{5, dx} \land \neg (A^{0}_{5, dx} \land Aspira^0)) \implies B^{1}_{5,dx} \qquad B^{0}_{5, dx} \land \neg (A^{0}_{5, dx} \land Aspira^0)}{B^{1}_{5,dx}}$


Si ottiene che *$B^{1}_{5,dx}$ = vero*

<br>

Transition 2:

dalla transition precedente si ha che KB $\cup$ {$A^{1}_{5,dx} \land \neg B^{1}_{5,sx} \land B^{1}_{5,dx}$}

Per rendere vero $Goal^2$ serve rendere $\neg B^{2}_{5,dx}$ vera e quindi $B^{2}_{5,dx}$ falso, per fare ciò bisogna far si che la parte destra dell'equivalenza logica sia falsa, quindi:

$B^{t+1}_{5,dx} \Leftrightarrow (B^{t}_{5, dx} \land \neg (A^{t}_{5, dx} \land Aspira^t))$

si nega la parte destra facendola diventare:

$\neg (B^{t}_{5, dx} \land \neg (A^{t}_{5, dx} \land Aspira^t))$

Con De Morgan diventa:

$\neg (B^{t}_{5, dx} \land \neg (A^{t}_{5, dx} \land Aspira^t)) \equiv (\neg B^{t}_{5, dx} \lor (A^{t}_{5, dx} \land Aspira^t))$

Avendo:

* $B^{1}_{5,sx}$ = Falso dalla transition precedente

* $A^{1}_{5,dx}$ = Vero dalla transition precedente

Per rendere vero tutto l'assioma $Aspira^1$ deve essere vero

Così facendo $B^{2}_{5,dx}$ = falso e $\neg B^{2}_{5,dx}$ = *vero*

<br>

da:

$B^{t+1}_{5,sx} \Leftrightarrow (B^{t}_{5, sx} \land \neg (A^{t}_{5, sx} \land Aspira^t))$

come effettuato in precedenza si ottiene:

$B^{t+1}_{5,sx} \implies (B^{t}_{5, sx} \land \neg (A^{t}_{5, sx} \land Aspira^t))$

che per contraddizione diventa:

$\neg (B^{t}_{5, sx} \land \neg (A^{t}_{5, sx} \land Aspira^t)) \implies \neg B^{t+1}_{5,sx}$

Visto che $B^{1}_{5,sx}$ = falso come si è visto in precedenza si ottiene che:

$B^{1}_{5, sx} \land \neg (A^{1}_{5, sx} \land Aspira^1)$ = falso

essendo però tutto negato, l'assioma diventa vero, ciò ci permette di applicare modus ponens:

$\frac{\neg (B^{1}_{5, sx} \land \neg (A^{1}_{5, sx} \land Aspira^1)) \implies \neg B^{2}_{5,sx} \qquad \neg (B^{1}_{5, sx} \land \neg (A^{1}_{5, sx} \land Aspira^1))}{\neg B^{2}_{5,dx}}$

ottenendo che:

$\neg B^{2}_{5,dx}$ è vera.

<br>

Avendo ottenuto $\neg B^{2}_{5,sx}$ e $\neg B^{2}_{5,dx}$ entrambe vere, il goal è stato raggiunto e quindi:

KB $\land Init^0 \land$ Transition 1 $\land$ Transition 2 $\land Goal^2 \models (Destra^0 \land Aspira^1)$

**Pianificare sfruttando un algoritmo SAT**

Per trovare una soluzione al problema si pianifica il tutto sfruttando SATPLAN, secondo la seguente serie di passaggi:

* Si costruisce un'unica formula formata dalla congiunzione di: $Init^0 \land Transition 1 \land \cdots \land Transition T \land goal^T$

* Si trasforma questa formula in CNF

* Si utilizza SATPLAN che invoca SAT-SOLVER:
    * Se non esiste alcun modello, il gioco non è risolvibile in n mosse: n incrementa di 1 e si ripete
    * Appena viene trovata una soluzione, questa è di lunghezza minima

* Si estraggono in ordine di transizione le variabili a cui è stato assegnato true e quella sequenza costituirà il piano di risoluzione per l'agente

## Esercizio 5 · Dimostrazione di un teorema

Dimostrare la validità del seguente <span style="color:#d4a017">teorema</span>, data la KB di partenza:

**KB:**

1. *Giuseppe suona con Gianni.*
2. *Gianni è un bassista oppure è un batterista.*
3. *Se Gianni fosse un batterista, allora sarebbe diplomato al conservatorio.*
4. *Giuseppe è diplomato al conservatorio.*
5. *Gianni non è diplomato al conservatorio.*

<span style="color:#d4a017">**Tesi ($\alpha$):**</span> *Gianni è un bassista.*

**Risposta**

GGiu = Giuseppe suona con Gianni

GBass = Gianni è un bassista

GBatt = Gianni è un batterista

GC = Gianni è diplomato al conservatorio

GiuC = Giuseppe è diplomato al conservatorio

KB:{$GGiu, GBass \lor GBatt, GBatt \implies GC, \neg GC, GiuC$}

KB \models \alpha

\alpha = GBass

<br>
Per contrapposizione di $GBatt \implies GC$ si ottiene:

$\neg GC \implies \neg GBatt$

Poichè $\neg GBatt$ è vero dalla base di conoscenza, per modus ponens si ottiene:

$\frac{\neg GC \implies \neg GBatt \qquad \neg GC}{\neg GBatt}$

Quindi $\neg GBatt$ è vero.

<br>

Poiché Gbatt è falso, visto che $\neg GBatt$ è vero, allora per far si che $GBass \lor GBatt$ sia vera: *GBass è vera*

quindi $\alpha$ = GBass è conseguenza logica di KB.
