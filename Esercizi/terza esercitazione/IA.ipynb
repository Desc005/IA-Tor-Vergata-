## BFS (Breadth-First Search)

L'algoritmo BFS è uno degli algoritmi base nella ricerca all'interno di grafi/alberi in quanto offre la sicurezza che se dovesse esistere una soluzione o un nodo goal questo venga raggiunto diventando quindi un algoritmo **COMPLETO**, ciò però solamente nei casi in cui ogni passaggio tra nodi abbia lo stesso costo di tutti gli altri o non abbia costo e se il numero di nodi ad ogni livello sia costituito da un numero _b_ finito.

Per quale motivo ciò?

Essendo l'algoritmo strutturato in modo da gestire la frontiera con una coda FIFO, questo esplorerà iterativamente tutti i livelli del grafo inserendo i nodi di ogni livello nella coda: esso esplorerà prima la radice (profondità 0), successivamente esplorerà tutti i _b_ nodi del livello 1, successivamente tutti i nodi _$b^2$_ del secondo livello, continuando così fino a raggiungere il goal. Poiché si presuppone che ogni livello contenga un numero finito di nodi (_$b^k$_ al livello _k_), se il goal esiste e si trova ad una profondità finita _d_, l'algoritmo esaminerà in tempo finito tutti i nodi precedenti (livelli 0, 1, ..., _d_-1) arrivando ad esaminare il nodo goal al livello _d_ senza arrivare a perdersi in rami infiniti con cammini _m_=$\infty$

Il codice sottostante mostra come nella coda dei nodi da esplorare vengano inseriti prima tutti i nodi del livello successivo a quello corrente del nodo esplorato (questi verranno esplorati solo successivamente che la coda si sia liberata di quelli dell'attuale livello).


Per quale motivo gli archi devono avere tutti costi uguali?

A differenza di algoritmi quali Dijkstra, Bellman-Ford o A*, i quali calcolano il percorso fino al nodo goal scegliendo quello di costo minore in grafi pesati, l'algoritmo BFS per avere un comportamento ottimale necessita che il costo di ogni arco del grafo sia costante con _k_ > 0, avendo quindi:

##### _g(n)_ = _k_ * depth(_n_)
###### (il costo per arrivare ad n sia una funzione crescente in quanto tutti gli archi hanno costo costante)

Questo comportamento risulta sfavorevole quando in un grafo pesato si ha un nodo intermedio, che collega radice e nodo goal ma che presenta un costo per percorrere l'arco che porta al goal _g_(goal) enorme (es. _g_(goal) = 100), e una serie di nodi che in alternativa porterebbero dalla radice al nodo goal ma con un costo totale di percorrenza g(goal) nettamente inferiore (es. _g_(goal) = 45). In questa casistica l'algoritmo BFS troverebbe una soluzione percorrendo il singolo nodo di costo enorme ottenendo così una soluzione a profondità 2 anziché quella ottima a profondità maggiore. Questo perché utilizzando una coda FIFO e non tenendo conto dei costi degli archi la BFS effettua un'esplorazione crescente per profondità inserendo nel cammino il primo nodo a livello _d_ incontrato senza tener conto dei costi.

### OTTIMALITÀ

Per essere definito ottimo, un algoritmo deve poter trovare una soluzione tale che _g(goal)=$C^*$_ minimo, cioè, la soluzione trovate deve avere costo minore possibile. Seppur la BFS sia completa, riuscendo a trovare sempre un cammino da un nodo radice ad un nodo goal, essa presenta un grosso limite legato alla tipologia del grafo. Infatti, se il grafo dovesse avere costi uniformi degli archi, la BFS riuscirebbe a trovare un cammino dal nodo radice al nodo goal con un costo proporzionale al numero di archi percorsi _g(n) = k * depth(n)_, grazie all'utilizzo di una coda FIFO che permette l'esplorazione del grafo per livelli di profondità _d_ (esplorando quindi tutti i nodi di un livello prima di passare al prossimo), ottenendo così un percorso ottimo e risultando quindi un algoritmo ottimale. Se invece si dovesse avere un grafo con costi eterogenei per ogni arco, la BFS non riuscirebbe a trovare il percorso ottimo (come nell esempio precedente potrebbe percorrere un percorso con meno archi ma con costo maggiore invece di prediligere quello con costo minore che attraversa più archi) risultando quindi un algoritmo non sempre ottimo.

### COMPLESSITÀ (TEMPORALE E SPAZIALE)

#### TEMPORALE
L'algoritmo BFS si struttura in modo da partire dal nodo radice e generare i nodi livello per livello: La radice dell'albero di ricerca genererà _b_ nodi, ognuno dei quali genererà altri b nodi per un totale di _$b^2$_ e così via. Avendo la soluzione a profondità _d_ arriveremo ad avere un numero di nodi generati pari a _$b^d$_.

###### **_(Fonte utilizzata: Artificial Intelligence A Modern Approach Fourth Edition, pagina 95)_**

> Domanda: il tempo per la generazione di un nodo è O(1)?
>
> Risposta: Sì, nell'analisi teorica standard degli algoritmi di ricerca in Intelligenza Artificiale si assume convenzionalmente che il tempo per generare un singolo nodo sia
O(1).



Poiché il tempo per la generazione di un singolo nodo è pari a O(1), si ottiene che la generazione di tutti i nodi richiede tempo esponenziale pari a:

##### $$T(b,d) = b + b^2 + \dots + b^d = \sum_{i=1}^d b^i = b \left( \frac{b^d - 1}{b - 1} \right)$$

Poiché il termine di grado massimo prevale asintoticamente sull'intera sommatoria, si ottiene che:
$$T(b, d) \in O(b^d)$$

Parallela alla complessità O(_n_ + _m_) utilizzata in grafi già generati e nel corso di ASD:



> ***Conversazione con l'IA Gemini Argon***

> Domanda: nella lezione di algoritmi e strutture dati ho studiato che la BFS ha un complessità temporale pari a O(n+m), perché qui si dice che la complessità è invece O(n^d)?
>
> Risposta: Le due notazioni esprimono la stessa complessità in due contesti diversi: in ASD il grafo è statico e già allocato in memoria ($n$ nodi, $m$ archi). In IA lo spazio degli stati è implicito e viene generato al volo. Nel sottografo esplorato fino a profondità $d$, il numero di vertici generati è $n \approx b^d$ e gli archi esaminati sono $m \approx b \cdot n \approx b^{d+1}$. Sostituendo nella formula di ASD si ottiene $O(n + m) = O(b^d + b^{d+1}) = O(b^d)$.




#### SPAZIALE
La complessità spaziale determina quanto spazio occupa in memoria l'esecuzione di un algoritmo, l'utilizzo di una coda FIFO come frontiera, la quale permette un'esplorazione livello per livello, farà si che arrivati all'ultimo livello _d_ essa conterrà tutti i _$b^d$_ nodi.


> ***Conversazione con l'IA Gemini Argon***
>
> Domanda: Leggendo lo pseudo-codice del BFS sembra che vengano espansi i nodi fino al livello _d_, mi sbaglio?
>
> Risposta: No, i nodi effettivamente espansi (estratti dalla frontiera e memorizzati nella lista dei nodi esplorati) appartengono solo ai livelli interni da $0$ a $d-1$. I nodi di livello $d$ non vengono espansi: poiché il goal-test nella BFS ha successo non appena il nodo obiettivo viene generato, l'esecuzione termina immediatamente lasciando i nodi di livello $d$ nella frontiera.


Esplorando un intero albero, verranno espansi i nodi dei livelli 0 fino al livello d-1 facendo in modo che nella frontiera FIFO vengano inseriti i nodi fino al livello _d_. La memoria richiesta quindi viene dominata dal numero di nodi presenti nella frontiera, il quale, essendo i nodi fino al livello _d_, farà avere alla BFS una complessità spaizale pari a:

##### $S(b, d) = O(b^d)$

La quale come per la complessità temporale risulta essere esponenziale alla profondità del nodo goal.

## RICERCA IN PROFONDITÀ LIMITATA E INCREMENTAL DEEPENING

L'algoritmo della *ricerca in profondità limitata* è una variante dell'algoritmo della ricerca in profondità che fissa come obiettivo quello di risolvere il problema della navigazione di un cammino di lunghezza infinita _m_ = $\infty$. Ciò viene risolto applicando un livello limite di profondità _***l***_ e trattando tutti i nodi a quel livello come se non avessero successori. Per evitare di imbattersi in cicli durante l'esplorazione, durante l'analisi dei nodi si guardano pochi collegamenti prima del padre in modo da scovare cicli, per cicli più lunghi il limite di esplorazione li eviterà. Per la stragrande maggioranza dei problemi non si può stabilire un buon livello _l_ fin quando non si risolve il problema. L'algoritmo *incremental deepening* risolve il problema provando tutti i valori possibili di _l_ fin quando non viene trovata una soluzione o ritorna un valore di fallimento.

### COMPLETEZZA

Entrambi gli algoritmi utilizzano uno stack LIFO il quale permette loro di analizzare in profondità ogni nodo, essi infatti analizzeranno prima tutti i nodi di ogni ramo fino a quando, arrivati al nodo a profondità _m_ massima, inizieranno ad analizzare tutti i nodi del ramo successivo e così via. Per essere definito completo un algoritmo deve trovare sempre una soluzione, in questo caso un percorso dal nodo radice al nodo goal. Se si esamina la ricerca in profondità si può notare come per limitazioni di profondità _l_ sfavorevoli, nell'eventualità che il nodo goal sia ad una profondità _d_ > _l_ questo non arriverà mai al nodo goal, confermando la non completezza dell'algoritmo.
