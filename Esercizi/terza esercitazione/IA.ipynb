### BFS (Breadth-First Search)

L'algoritmo BFS è uno degli algoritmi base nella ricerca nei grafi/alberi in quanto offre la sicurezza che se dovesse esistere una soluzione o un nodo goal questo viene raggiunto divenatando quindi un algortimo **COMPLETO**, ciò però solo nei casi in cui ogni passaggio tra nodi abbia lo stesso costo di tutti gli altri o non abbia costo.

Per quale motivo ciò?

Essendo l'algoritmo strutturato in modo da esplorare in maniera ricorsiva (o iterativa, in base a come viene scritto l'algoritmo) tutti i nodi di ogni livello, esplorando quindi prima la radice, poi tutti i nodi del livello uno collegati alla ridice, successivamente tutti quelli del secondo livello e così via, ci offre l sicurezza che esplorando ogni singolo nodo riesca ad arrivare con certezza al nodo goal se questo esiste o è raggiungibile

Il codice sottostante mostra come nella coda dei nodi da esplorare vengano inseriti prima tutti i nodi del livello successivo a quello corrente adiacenti al nodo esplorato (questi verranno esplorati solo successivamente che la coda si sia liberata di quelli dell'attuale livello).
