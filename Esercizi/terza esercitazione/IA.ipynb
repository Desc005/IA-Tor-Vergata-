# Discussione algoritmidi ricerca

Durante le lezioni del corso di Intelligenza artificiale sono stati trattati diversi algoritmi per la ricerca in alberi/grafi alcuni dei quali sono:
* BFS (Breadth-First Search)
* DFS (Depth-First Search)
* Ricerca in profondità limitata - ID (Incremental Deepening)
* Ricerca Bidirezionale

Trattandone le loro esattezze e completezze, le quali verranno discusse in questo notebook, assumendo come a lezione che:
* b = fattore di ramificazione (branching) (numero max di successori)
* d = profondità del nodo obiettivo più superficiale
* m = lunghezza massima dei cammini nello spazio di ricerca.
* k = costo degli archi
* g(n) = costo del cammino per arrivare ad n

## BFS (Breadth-First Search)

L'algoritmo BFS è uno degli algoritmi base nella ricerca all'interno di grafi/alberi in quanto offre la sicurezza che se dovesse esistere una soluzione o un nodo goal questo venga raggiunto diventando quindi un algoritmo **COMPLETO**, ciò però solamente nei casi in cui ogni passaggio tra nodi abbia lo stesso costo di tutti gli altri o non abbia costo e se il numero di nodi ad ogni livello sia costituito da un numero _b_ finito.

Per quale motivo ciò?

Essendo l'algoritmo strutturato in modo da gestire la frontiera con una coda FIFO, questo esplorerà iterativamente tutti i livelli del grafo inserendo i nodi di ogni livello nella coda: esso esplorerà prima la radice (profondità 0), successivamente esplorerà tutti i _b_ nodi del livello 1, successivamente tutti i nodi _$b^2$_ del secondo livello, continuando così fino a raggiungere il goal. Poiché si presuppone che ogni livello contenga un numero finito di nodi (_$b^k$_ al livello _k_), se il goal esiste e si trova ad una profondità finita _d_, l'algoritmo esaminerà in tempo finito tutti i nodi precedenti (livelli 0, 1, ..., _d_-1) arrivando ad esaminare il nodo goal al livello _d_ senza arrivare a perdersi in rami infiniti con cammini _m_=$\infty$


Per quale motivo gli archi devono avere tutti costi uguali?

A differenza di algoritmi quali Dijkstra, Bellman-Ford o A*, i quali calcolano il percorso fino al nodo goal scegliendo quello di costo minore in grafi pesati, l'algoritmo BFS per avere un comportamento ottimale necessita che il costo di ogni arco del grafo sia costante con _k_ > 0, avendo quindi:

##### _g(n)_ = _k_ * depth(_n_)
###### (il costo per arrivare ad n sia una funzione crescente in quanto tutti gli archi hanno costo costante)

Questo comportamento risulta sfavorevole quando in un grafo pesato si ha un nodo intermedio, che collega radice e nodo goal ma che presenta un costo per percorrere l'arco che porta al goal _g_(goal) enorme (es. _g_(goal) = 100), e una serie di nodi che in alternativa porterebbero dalla radice al nodo goal ma con un costo totale di percorrenza g(goal) nettamente inferiore (es. _g_(goal) = 45). In questa casistica l'algoritmo BFS troverebbe una soluzione percorrendo il singolo nodo di costo enorme ottenendo così una soluzione a profondità 2 anziché quella ottima a profondità maggiore. Questo perché utilizzando una coda FIFO e non tenendo conto dei costi degli archi la BFS effettua un'esplorazione crescente per profondità inserendo nel cammino il primo nodo a livello _d_ incontrato senza tener conto dei costi.

### OTTIMALITÀ

Per essere definito ottimo, un algoritmo deve poter trovare una soluzione tale che _g(goal)=$C^*$_ minimo, cioè, la soluzione trovate deve avere costo minore possibile. Seppur la BFS sia completa, riuscendo a trovare sempre un cammino da un nodo radice ad un nodo goal, essa presenta un grosso limite legato alla tipologia del grafo. Infatti, se il grafo dovesse avere costi uniformi degli archi, la BFS riuscirebbe a trovare un cammino dal nodo radice al nodo goal con un costo proporzionale al numero di archi percorsi _g(n) = k * depth(n)_, grazie all'utilizzo di una coda FIFO che permette l'esplorazione del grafo per livelli di profondità _d_ (esplorando quindi tutti i nodi di un livello prima di passare al prossimo), ottenendo così un percorso ottimo e risultando quindi un algoritmo ottimale. Se invece si dovesse avere un grafo con costi eterogenei per ogni arco, la BFS non riuscirebbe a trovare il percorso ottimo (come nell esempio precedente potrebbe percorrere un percorso con meno archi ma con costo maggiore invece di prediligere quello con costo minore che attraversa più archi) risultando quindi un algoritmo non sempre ottimo.

La quale come per la complessità temporale risulta essere esponenziale alla profondità del nodo goal.

## RICERCA IN PROFONDITÀ LIMITATA E INCREMENTAL DEEPENING

L'algoritmo della *ricerca in profondità limitata* è una variante dell'algoritmo della ricerca in profondità che fissa come obiettivo quello di risolvere il problema della navigazione di un cammino di lunghezza infinita _m_ = $\infty$. Ciò viene risolto applicando un livello limite di profondità _***l***_ e trattando tutti i nodi a quel livello come se non avessero successori, a differenza del DFS classico questa operazione richiede tre tipi di ritorno:
* Soluzione (se il nodo viene trovato)
* Cutoff (se si è raggiunto il limite _l_ stabilito)
* Fallimento (se si è esplorato ogni ramo fino al limite _l_ senza trovare la soluzione)

Per evitare di imbattersi in cicli durante l'esplorazione, durante l'analisi dei nodi si guardano pochi collegamenti prima del padre in modo da scovare cicli, per cicli più lunghi il limite di esplorazione li eviterà. Per la stragrande maggioranza dei problemi non si può stabilire un buon livello _l_ fin quando non si risolve il problema. L'algoritmo *incremental deepening* risolve il problema provando tutti i valori possibili di _l_ fin quando non viene trovata una soluzione o ritorna un valore di fallimento.

### COMPLETEZZA

Entrambi gli algoritmi utilizzano uno stack LIFO il quale permette loro di analizzare in profondità ogni nodo, essi infatti analizzeranno prima tutti i nodi di ogni ramo fino a quando, arrivati al nodo a profondità _l_, inizieranno ad analizzare tutti i nodi del ramo successivo e così via. Per essere definito completo un algoritmo deve trovare sempre una soluzione, in questo caso un percorso dal nodo radice al nodo goal. Se si esamina la ricerca in profondità si può notare come per limitazioni di profondità _l_ sfavorevoli, nell'eventualità che il nodo goal sia ad una profondità _d_ > _l_, questo non arriverà mai al nodo goal, confermando la non completezza dell'algoritmo.

Incremental deepening risolve questa non completezza, in quanto, provando tutti i possibili valori di _l_ fin a che non venga trovata una soluzione o si fallisca, garantisce che, se dovesse esistere, il nodo goal a profondità _d_ venga raggiunto.

È completo in tutti i casi?
Così come per la BFS anche Incremental deepening per essere completo richiede che il numero di nodi ad ogni livello sia in numero finito (_b_ sia finito), così da non esplorare un numero infinito di branch (data la sua esplorazione in profondità). Inoltre sempre come la BFS, Iterative deepening è completo se lo spazio di esplorazione contiene la soluzione oppure è finito (per eveitare un'esplorazione infinita).

Se l'algoritmo non è completo, significa che in determinati grafi o configurazioni non riuscirà a trovare alcuna soluzione, fallendo per definizione la garanzia di restituire l'ottimo.


Per essere ottimale, un algoritmo deve trovare la soluzione di costo minimo _g(goal)=$C^*$_ qualora questa esista. Per essere ottimo un algoritmo necessita quindi di essere completo, essendo la ricerca in profondità limitata un algoritmo non completo (a causa di configurazioni sfavorevoli di _l_ che porterebbero a scartare il livello con il nodo goal o a scoprire un goal a profondità _l_ prima di un altro a profondità _d_ < _l_), si ottiene di conseguenza che non è un algoritmo ottimo.

Discorso opposto riguarda Incremental deepening, così come per la BFS e per la ricerca in profondità limitata anche Incremental deepening richiede che i costi per passare da un nodo ad un altro siano costanti _g(n) = k * depth(n)_. La sua ottimalità infatti dipende proprio da ciò, in quanto, non tenendo conto dei costi, potrebbe prediligere un cammino che porta al nodo goal di costo enorme ma che attraversa pochi nodi, invece di prediligere quello di costo minore ma che attraversa più nodi, fornendo una soluzione non ottima.

Se i costi sono costanti _g(n) = k * depth(n)_?

Così come per la DFS, grazie allo stack LIFO utilizzato come frontiera, esplorerà tutti i nodi del grafo analizzando prima tutti quelli di un ramo per poi passare a quelli del ramo successivo e così via. A differenza della DFS classica qui viene introdotto il fattore _l_ indicante la profondità massima a cui arrivare per analizzare i nodi. Questo fattore però non influisce sull'ottimalità di Incremental deepening, in quanto, calcolando in maniera iterativa ogni valore di _l_ fino a trovare quello ottimo che comprende il nodo goal (_d_ = _l_), fa si che venga sempre trovata la soluzione e che, provando ogni profondità per ogni valore di _l_, sia anche quella ottima.

#### TEMPORALE

RICERCA IN PROFONDITÀ LIMITATA

Come la DFS classica la cui complessità era dettata dalla cammino massimo sul grafo finendo per offrire una complessità temporale di $O(b^m)$, la ricerca in profondità limitata segue una logica simile, con l'unica peculiarità che non si raggiunge la profondità massima del grafo, ma si viene limitati ad una profondità pari ad _l_. Essa infatti esplorerà prima la radice, poi esplorerà il figlio, successivamente il figlio del figlio e così via fino a raggiungere una profondità di _l_ dopo la quale non esplorerà più in profondità. Una volta terminata l'esplorazione del nodo a quel livello, l'algoritmo esplorerà i restanti fratelli _b_ generati dallo stesso padre al livello _l_ prima di tornare al livello precedente per esplorare gli altri rami, continuando così fin quando non si è esaurito il grafo o si è raggiunto il goal.

Poiché ogni esplorazione su un nodo ha costo costante O(1), si ottiene che l'esplorazione di tutti i nodi richiede:

$$T(b,l) = b^l + b^{l-1} + \dots + b^1 = \sum_{i=1}^l b^i = b \left( \frac{b^l - 1}{b - 1} \right)$$

poiché il termine di grado massimo prevale nella sommatoria, si ottiene che:

$$T(b, l) \in O(b^l)$$

Il che risulta esponenziale sulla profondità massima raggiungibile _l_.

INCREMENTAL DEEPENING

Così come descritto per la ricerca in profondità limitata l'Incremental deepening effettua le stesse medesime operazione per quanto riguarda l'esplorazione dei nodi, l'unica differenza è l'aumento iterativo del fattore _l_ per ottenere il livello ottimale che includa il nodo goal (_l_ = _d_). Pur di salvare dello spazio, l'algoritmo sacrifica delle prestazioni in fatto di tempo, in quanto a differenza della BFS, l'Incremental deepening non salava tutti i nodi precedenti a quelli nuovi da generare in memoria, ma li rigenera da capo. La creazione di nuovi livelli con questa metodologia ad ogni iterazione di Incremental deepening può sembrare uno spreco, in quanto i livelli vicino la radice vengono generati più volte, ma per molti problemi, la maggior parte dei nodi si trova verso le foglie, non importa che i nodi iniziali vengano rigenerati.

Dato che la generazione dei nodi al livello _d_ avviene una sola volta, quelli a livello precedente due volte, e così via, l'Incremental deepening presenta la seguente serie:

$$T(b,d) = (d)b + (d-1)b^2 + \dots + b^d = \sum_{i=1}^d (d-i+1)b^i$$

Risolvendo la sommatoria aritmetico-geometrica, il termine di ordine massimo rimane dominato da $b^d$, ottenendo:

$$T(b, d) \in O(b^d)$$

> **Esempio con BFS (Artificial Intelligence A Modern Approach pagina 99):**
> Con fattore di ramificazione $b = 10$ e profondità $d = 5$:
> * $N(\text{BFS}) = 10 + 100 + 1.000 + 10.000 + 100.000 = 111.110$ nodi
> * $N(\text{ID}) = 50 + 400 + 3.000 + 20.000 + 100.000 = 123.450$ nodi
> Essendo la differenza quasi minima alla BFS, viene conservata la medesima classe di complessità temporale asintotica $O(b^d)$.

Il che risulta esponenziale sulla profondità del nodo goal _d_.

> Domanda: ma se lo stack contiene tutti i nodi da esplorare oltre quelli del cammino che si sta facendo ora, non dovrebbe essere O(b^l) la complessità spaziale?
>
> Risposta: No, ed è proprio questo il motivo per cui la ricerca in profondità (e quindi DLS e ID) è così celebre per il risparmio di memoria: la complessità spaziale rimane rigorosamente lineare, ossia O(b⋅l), e non esponenziale $O(b^l)$ come accade nella BFS. La differenza fondamentale deriva dal fatto che lo stack non contiene mai tutti i nodi dell'albero, ma unicamente i nodi presenti lungo il singolo ramo attivo e i relativi fratelli non ancora esplorati ad ogni livello di quel ramo.


Come per la DFS classica, l'utilizzo di uno stack come frontiera permette all'algoritmo di tenere in memoria solo i nodi esplorati dalla radice fino al nodo corrente e tutti i nodi fratelli ancora non esplorati, grazie a questa proprietà si fa in modo che nello stack non siano mai presenti tutti i nodi del grafo, ma solo quelli necessari per l'esplorazione in profondità. Dato che la dimensione massima dello stack si riscontra al livello finale _l_, il numero di nodi al suo interno sarà uguale a $(b−1)⋅l+1≈b⋅l$.

Data la formula precedente si può concludere che la complessità spaziale è lineare rispetto al livello di profondità _l_ stabilito (dato che _b_ è costante):

$$S(b,l) = O(bl)$$

INCREMENTAL DEEPENING

Poiché Incremental deepening non salva in memoria i nodi precedentemente generati, rigenerandoli ad ogni iterazione, il suo comportamento per ogni valore di _l_ è esattamente quello di una ricerca in profondità limitata con _l_ che varia ad ogni iterazione. Ad ogni fallimento o cutoff, infatti, lo stack viene completamente liberato, permettendone il riuso, così che si possano generare nuovamente i nodi incrementando _l_ di 1. Dato che l'incremento di _l_ si interrompe al raggiungimento del livello contenente il nodo goal (_l_ = _d_), oppure all'arrivo della profondità massima, si ha che il numero di nodi massimo all'interno dello stack sarà uguale a $(b−1)⋅d+1≈b⋅d$

Data la formula precedente si può concludere che la complessità spaziale è lineare rispetto al livello di profondità _d_ del nodo goal:

$$S(b,d) = O(bd)$$

Questo livello di efficienza rende questo algoritmo più performante nello spazio rispetto all'algoritmo BFS analizzato precedentemente.
